```python
from openpyxl import Workbook
from datetime import datetime

wb = Workbook()
ws = wb.active

today = datetime.today()

ws['B2'] = "請求書"
ws['B4'] = "株式会社ABC"
ws['B5'] = "〒101-0022 東京都千代田区神田練塀町300"
ws['B6'] = "TEL:03-1234-5678 FAX:03-1234-5678"
ws['B7'] = "担当者名:鈴木一郎 様"
ws['F4'] = "No."
ws['G4'] = "0001"
ws['F5'] = "日付"
ws['G5'] = f"{today.strftime('%Y/%m/%d')}"

ws['B10'] = "商品名"
ws['C10'] = "数量"
ws['D10'] = "単価"
ws['E10'] = "金額"

ws['B11'] = "商品A"
ws['C11'] = 2
ws['D11'] = 10000
ws['E11'] = ws['C11'].value * ws['D11'].value

ws['B12'] = "商品B"
ws['C12'] = 1
ws['D12'] = 15000
ws['E12'] = ws['C12'].value * ws['D12'].value

ws['E13'] = ws['E11'].value + ws['E12'].value

ws['B15'] = "合計"
ws['E15'] = ws['E13'].value

ws['B16'] = "消費税"
ws['E16'] = ws['E15'].value * 0.1

ws['B17'] = "税込合計"
ws['E17'] = ws['E15'].value + ws['E16'].value

backup_filename = f"請求書_{today.strftime('%Y%m%d')}.xlsx"
wb.save(backup_filename)
```
